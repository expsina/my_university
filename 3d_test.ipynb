```python
import numpy as np
import plotly.graph_objects as go

# 1. Создаем координаты X и Y
x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)
X, Y = np.meshgrid(x, y)

# 2. Вычисляем Z (функция волны)
Z = np.exp(X + Y)

# 3. Создаем 3D-поверхность
fig = go.Figure(data=[go.Surface(x=x, y=y, z=Z, colorscale='Viridis')])

# 4. Настраиваем подписи осей и заголовок
fig.update_layout(
    title='Интерактивная 3D-поверхность в Plotly',
    scene=dict(
        xaxis_title='Ось X',
        yaxis_title='Ось Y',
        zaxis_title='Ось Z'
    ),
    margin=dict(l=0, r=0, b=0, t=40) # Убираем лишние отступы по краям
)

# 5. Отображаем график в браузере или Jupyter
fig.show()
```

```python
import numpy as np
import plotly.graph_objects as go

# 1. Задаем параметры тора
R = 3  # Расстояние от центра трубы до центра тора (большой радиус)
r = 1  # Радиус самой трубы (малый радиус)

# 2. Генерируем сетку углов (от 0 до 2*pi)
u = np.linspace(0, 2 * np.pi, 100)
v = np.linspace(0, 2 * np.pi, 100)
u, v = np.meshgrid(u, v)

# 3. Параметрические уравнения тора
x = (R + r * np.cos(v)) * np.cos(u)
y = (R + r * np.cos(v)) * np.sin(u)
z = r * np.sin(v)

# 4. Создаем 3D-поверхность в Plotly
fig = go.Figure(
    data=[
        go.Surface(
            x=x,
            y=y,
            z=z,
            colorscale="Viridis",  # Цветовая схема (можно заменить на 'Plasma', 'Cividis' и др.)
            colorbar=dict(title="Высота Z"),  # Боковая панель цвета
        )
    ]
)

# 5. Настраиваем отображение осей (сохраняем пропорции 1:1:1)
fig.update_layout(
    title="3D Тор в Plotly",
    scene=dict(
        xaxis=dict(range=[-5, 5]),
        yaxis=dict(range=[-5, 5]),
        zaxis=dict(range=[-5, 5]),
        aspectmode="cube",  # Делает оси равными, чтобы тор не искажался
    ),
    margin=dict(l=0, r=0, b=0, t=40),  # Убираем лишние отступы
)

# 6. Отображаем график
fig.show()
```
