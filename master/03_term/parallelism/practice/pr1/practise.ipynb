```python
import threading
import numpy as np
```

```python
max_number = 0
lock       = threading.Lock()
```

```python
def find_max(ar, start, end):
    global max_number 

    local_max_number = 0

    for i in range(start, end):
        if ar[i] > local_max_number:
            local_max_number = ar[i]

    with lock:

        if local_max_number > max_number:
            max_number = local_max_number
```

```python
threads = []
num_threads = 2

size = 100000000
np.random.seed(42)
data = np.random.rand(size)

chunk_size = size // num_threads

for i in range(num_threads):
    start = i * chunk_size 
    end = size if i == num_threads - 1 else (i + 1) * chunk_size 

    thread = threading.Thread(target=find_max, args=(data, start, end))

    threads.append(thread)
    thread.start()

for t in threads:
    t.join()

print(f"Get from threads max_number: {max_number}, max number in fact: {max(data)}")
```

```python
size = 10

final_arr = np.zeros(size)
```

```python
final_arr
```

```python
def find_sum_of_array(a, b, start, end):
    global final_arr
    output_arr = []

    for i in range(start, end):
        output_arr.append(a[start: end] + b[start:end])

    final_arr[start:end] = output_arr
```

```python
np.random.seed(52)
a = np.random.rand(size)
b = np.random.rand(size)

threads = []
num_threads = 2
chunk_size = size // num_threads

for i in range(num_threads):

    start = i * chunk_size
    end   = size if i == (num_threads - 1) else chunk_size * (i + 1) 

    thread = threading.Thread(target=find_sum_of_array, args=(a, b, start, end))

    threads.append(thread)
    thread.start()

for t in threads:
    t.join()
```
