```python
import threading
import time
```

```python
def calculate_fibonacci(n):

    fibonacci_numbers = [1, 1]

    for i in range(2, n):
        fibonacci_numbers.append(
            fibonacci_numbers[i-1]+fibonacci_numbers[i-2]
        )

    return fibonacci_numbers
```

```python
start_time = time.perf_counter()

calculate_fibonacci(100000)
calculate_fibonacci(100000)

single_thread_time = (time.perf_counter() - start_time)
single_thread_time
```

```python
start_time = time.perf_counter()

t1 = threading.Thread(target=calculate_fibonacci, args=(100000, ))
t2 = threading.Thread(target=calculate_fibonacci, args=(100000, ))

t1.start()
t2.start()
t1.join()
t2.join()

multi_thread_time = (time.perf_counter() - start_time)
multi_thread_time
```
