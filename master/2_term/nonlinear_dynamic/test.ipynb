```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Задаем временной отрезок t (в нашей системе всё растет быстро, возьмем малый интервал)
t = np.linspace(-1.5, 0.8, 1000)

# 2. Выбираем константы управления C1 и C2 (например, обе равны 1, как в обсуждении)
C1 = 1.0
C2 = 5.0

# 3. Формулы для x(t) и y(t), которые мы получили из фундаментальной матрицы
x = C1 * np.exp(4 * t) + C2 * np.exp(t)
y = C1 * np.exp(4 * t) - 2 * C2 * np.exp(t)

# --- Визуализация ---
plt.figure(figsize=(8, 8))

# Рисуем координатные оси
plt.axhline(0, color='black', linewidth=0.5, linestyle='--')
plt.axvline(0, color='black', linewidth=0.5, linestyle='--')

# Строим линии собственных векторов для наглядности
# Собственный вектор 1: v1 = (1, 1) -> линия y = x
line_v1 = np.linspace(-5, 10, 100)
plt.plot(line_v1, line_v1, color='green', alpha=0.5, linestyle=':', label='Собственное направление v1 (λ=4)')

# Собственный вектор 2: v2 = (1, -2) -> линия y = -2x
line_v2 = np.linspace(-3, 5, 100)
plt.plot(line_v2, -2 * line_v2, color='red', alpha=0.5, linestyle=':', label='Собственное направление v2 (λ=1)')

# Рисуем результирующую траекторию движения (x против y, без оси t)
plt.plot(x, y, color='blue', linewidth=2.5, label='Траектория движения (x, y)')

# Отмечаем начальную точку при t = 0 (x = 1+1=2, y = 1-2=-1)
x_start = C1 * np.exp(0) + C2 * np.exp(0)
y_start = C1 * np.exp(0) - 2 * C2 * np.exp(0)
plt.scatter([x_start], [y_start], color='darkblue', zorder=5)
plt.text(x_start + 0.2, y_start - 0.2, 'Точка при t = 0', fontsize=10, weight='bold')

# Стрелка, показывающая направление течения времени
plt.annotate('', xy=(x[-200], y[-200]), xytext=(x[-205], y[-205]),
             arrowprops=dict(arrowstyle="->", color="blue", lw=2.5))

# Настройка сетки и подписей
plt.title('Фазовая траектория системы (график X и Y без оси t)', fontsize=12)
plt.xlabel('Переменная X')
plt.ylabel('Переменная Y')
plt.xlim(-50, 50)
plt.ylim(-50, 50)
plt.grid(True, alpha=0.3)
plt.legend()

plt.show()
```
