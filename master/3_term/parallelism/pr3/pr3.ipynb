ЯПФ - это представление графа, где операции разбиты на ярусы (уровни, слои). Операции внутри одного яруса независимы и могут выполняться параллельно. Операции следующего яруса начинаются только после того, как весь предыдущий ярус завершен

```python
import numpy as np 

y = np.linspace(1, 100, 100)
z = np.linspace(1, 100, 100)

for i in range(3, 100):
    y[i] = y[i-3] + z[i]
```
