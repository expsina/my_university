```python
# 4.3.2
import math

diff_func = lambda x: math.sqrt(1 + math.pow(math.e, math.sqrt(x)) + math.cos(x ^ 2)) / (math.fabs(1 - math.pow(math.sin(x), 3))) + math.log1p(math.fabs(2 * x))

x = np.random.randint(1, 10, 10)
y = [diff_func(i) for i in x]

plt.grid()
plt.plot(x, y, c = "r")
```

```python
plt.scatter(x[:6], y[:6])
```

```python
from scipy.integrate import simpson

another_diff_func = lambda x: math.fabs(math.cos(x * math.pow(math.e, math.cos(x) + math.log1p(x + 1))))

x = np.arange(0, 11, 1)
y = [another_diff_func(i) for i in x]

plt.grid()
plt.plot(x, y, c = "r")
plt.fill_between(x, y)

area = np.trapz(y)
print(area)
```

```python
# 4.3.5

def plus(x, y):
    return x + y

def minus(x, y):
    return x - y

def multiply(x, y):
    return x * y

def divide(x, y):
    return x / y

def first_f(x, y):
    return math.pow(math.e, x + y)

def second_f(x, y):
    return math.sin(x + y)

def third_f(x, y):
    return math.cos(x + y)

def fourth_f(x, y):
    return math.pow(x, y)

x, y = map(int, input().split())

command = input()

calculator = {
    "+": plus(x, y),
    "-": minus(x, y),
    "*": multiply(x, y),
    "/": divide(x, y),
    "1": first_f(x, y),
    "2": second_f(x, y),
    "3": third_f(x, y),
    "4": fourth_f(x, y),
}

calculator[command]
```
