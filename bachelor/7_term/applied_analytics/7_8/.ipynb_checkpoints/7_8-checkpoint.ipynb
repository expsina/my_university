```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
ads = pd.read_csv("7_8/ads_yandex.csv")
calls = pd.read_csv("7_8/calls_new.csv")
sessions = pd.read_csv("7_8/sessions_new.csv")
```

```python
ads.head()
```

```python
count_ctr_cps = ads.groupby(by = ["ad_id"], dropna = True)[["cost", "clicks", "shows"]].sum()
count_ctr_cps.head()
```

```python
count_ctr_cps["ctr"] = count_ctr_cps["clicks"] / count_ctr_cps["shows"] * 100
count_ctr_cps["cps"] = count_ctr_cps["cost"] / count_ctr_cps["clicks"]
```

```python
count_ctr_cps[["ctr", "cps"]]
```

```python
plt.hist(count_ctr_cps["ctr"], bins = 20)
```
