```python
class Trigonometry:

    @staticmethod
    def fact(x):
        total = 1

        for i in range(1, x + 1):
            total *= i

        return total
    
    @staticmethod
    def radians(deg):
        return (deg * 3.14) / 180

    def __init__(self, deg):
        self.deg = deg
        self.rad = self.radians(deg)

    def sin(self):
        self.sx = 0

        for i in range(0, 10):
            self.sx += ( ((-1) ** i) * (self.rad ** (2 * i + 1))) / self.fact(2 * i + 1)

        return self.sx

    def cos(self):
        self.cs = 0

        for i in range(0, 10):
            self.cs += ((-1) ** i) * ( (self.rad ** (2 * i)) / self.fact(2 * i))

        return self.cs

    def tag(self):
        return self.cos() / self.sin()

    def arcsin(self):
        self.arcsin = 0

        for i in range(0, 10):
            self.arcsin += ((self.fact(2 * i)) / ((2 ** (2 * i)) * self.fact(i) ** 2)) * ((self.rad ** (2 * i + 1)) / (2 * i + 1))

        return self.arcsin

    def arccos(self):
        self.arccos = 0

        for i in range(0, 10):
            self.arccos += (3.14 / 2) - ((self.fact(2 * i)) / ((2 ** (2 * i)) * self.fact(i) ** 2)) * ((self.rad ** (2 * i + 1)) / (2 * i + 1))

        return self.arccos

    def arctag(self):
        self.arctag = 0

        for i in range(0, 10):
            self.arctag += (((-1) ** (i - 1)) * (self.rad ** (2 * i - 1))) / (2 * i -1)

        return self.arctag

new = Trigonometry(45)
new.arctag()
```

```python
class Tree:
  def __init__ (self, n):
    self.Name = n
    self.Left, self.Right = None, None
    
  def AddLeft(self, n):
    t = Tree(n)
    if self.Left == None:
      self.Left = t
    else:
      t.Left = self.Left
      self.Left = t
  
  def AddRight(self, n):
    t = Tree(n)
    if self.Right == None:
      self.Right = t
    else:
      t.Right = self.Right
      self.Right = t
  
  def Print(self, n):
    s = ' ' * n
    print(s + 'Имя корня: {0}'.format(self.Name))
    if self.Left != None:
      print(s + 'Левый потомок: {0}'.format(self.Left.Name))
    else:
      print(s + 'Левый потомок: ссылка равна nil.')
    
    if self.Right != None:
      print(s + 'Правый потомок: {0}'.format(self.Right.Name))
    else:
      print(s + 'Правый потомок: ссылка равна nil.')
  
  @classmethod
  def SeachBool(t, n):
    if t != None:
        if t.Name == n:
            return True
        else:
            return SeachBool(t.Left, n) or SeachBool(t.Right, n)
    else:
        return False
  
  def SeachSelf(self, n):
    SeachBool(self, t, n)
```

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split

df = [[-1, -1], [-2, -1], [-3, -2], [1, 1], [2, 1], [3, 2]]
target = [0, 0, 0, 1, 1, 1]
```

```python
for i in range(0, len(df)):
    df[i].append(target[i])

df = pd.DataFrame(df)
```

```python
from sklearn.tree import DecisionTreeClassifier

classifier = DecisionTreeClassifier()
classifier.fit(df.iloc[:, :-1], df.iloc[:, -1])
```

```python
from sklearn import tree
tree.plot_tree(classifier)
```

```python
df = pd.read_csv("https://raw.githubusercontent.com/likarajo/petrol_consumption/master/data/petrol_consumption.csv")
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    df.iloc[:, :-1],
    df.iloc[:, -1],
    test_size = 0.2,
    random_state = 0
)
```

```python
from sklearn.tree import DecisionTreeRegressor

regressor = DecisionTreeRegressor()
regressor.fit(X_train, y_train)
```

```python
tree.plot_tree(regressor)
```

```python
y_pred = regressor.predict(X_test)
y_pred
```

```python
df = pd.DataFrame({"Actual": y_test, "Predicted": y_pred})
df
```

```python
from sklearn import metrics
print("Mean Squared Error:", metrics.mean_squared_error(y_test, y_pred))
print("Mean Absolute Error:", metrics.mean_absolute_error(y_test, y_pred))
```
