```python
# 1.3.1

import numpy as np
import itertools

points = [np.random.randint(-10, 10, 3) for i in range(0, 4)]
i = 1


for a, b in itertools.combinations(points, 2):
    print(f"""
    {i}. {a} и {b}:\n
    Расстояние Евклида -- {np.linalg.norm(a - b)}\n
    Квадрат Евклидова расстояния -- {np.linalg.norm(a - b) * 2}\n
    Расстояние Чебышева -- {np.linalg.norm(a - b, ord = np.inf)}\n
    Расстояние Хемминга -- {np.linalg.norm(a - b, ord = 1)}
    """, end = "")
    i += 1
```

```python
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

fig = plt.figure()
ax = fig.add_subplot(111, projection = "3d")

for i in points:
    ax.scatter(*i)

plt.show()
```

```python
# 2.3.1

import seaborn as sns

iris = sns.load_dataset('iris')
```

```python
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

X_train, X_test, y_train, y_test = train_test_split(
    iris.iloc[:, :-1],
    iris.iloc[:, -1],
    test_size = 0.15
)

models = []

for i in (1, 5, 10):
    models.append(KNeighborsClassifier(n_neighbors = i))

for i in models:
    i.fit(X_train, y_train)

models_predictions = []

for i in models:
    models_predictions.append(i.predict(X_test))
```

```python
plt.figure(figsize = (10, 7))
sns.scatterplot(x = "petal_width", y = "petal_length",
               data = iris, hue = "species", s = 70)
plt.xlabel("Длина лепестка, см")
plt.ylabel("Ширина лепестка, см")
plt.legend(loc = 2)
plt.grid()
plt.title("N = 1")

for i in range(len(y_test)):
    if np.array(y_test)[i] != models_predictions[0][i]:
        plt.scatter(X_test.iloc[i, 3], X_test.iloc[i, 2],
                   color = "red", s = 150)

plt.show()
print(f"Accuracy: {accuracy_score(y_test, models_predictions[0]) :.3}")
```

```python
plt.figure(figsize = (10, 7))
sns.scatterplot(x = "petal_width", y = "petal_length",
               data = iris, hue = "species", s = 70)
plt.xlabel("Длина лепестка, см")
plt.ylabel("Ширина лепестка, см")
plt.legend(loc = 2)
plt.grid()
plt.title("N = 5")

for i in range(len(y_test)):
    if np.array(y_test)[i] != models_predictions[1][i]:
        plt.scatter(X_test.iloc[i, 3], X_test.iloc[i, 2],
                   color = "red", s = 150)

plt.show()
print(f"Accuracy: {accuracy_score(y_test, models_predictions[1]) :.3}")
```
