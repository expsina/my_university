```python
import matplotlib.pyplot as plt

# Фактические данные (2003-2023)
historical_years = list(range(2003, 2024))
historical_values = [64.7, 65.4, 66.0, 66.3, 67.1, 67.4, 67.6, 67.7, 68.3, 68.7, 68.5, 68.9, 69.1, 69.5, 69.0, 68.7, 67.8, 67.3, 67.8, 67.8, 68.5]

# Прогнозные годы (2024-2026)
years_forecast = [2024, 2025, 2026]

# Прогнозные значения уровня участия в рабочей силе (%)
forecast_values = [71.96, 72.23, 72.49]

# Нижняя и верхняя границы 95% доверительного интервала (%)
ci_lower = [70.49, 70.55, 70.60]
ci_upper = [73.43, 73.91, 74.38]

# Создание графика
plt.figure(figsize=(12, 7)) # Размер графика увеличен для лучшего отображения

# Построение линии фактических данных
plt.plot(historical_years, historical_values, marker='x', linestyle='-', color='green', label='Фактические данные (2003-2023)')

# Построение линии прогноза
plt.plot(years_forecast, forecast_values, marker='o', linestyle='-', color='blue', label='Прогноз (2024-2026)')

# Построение доверительного интервала
plt.fill_between(years_forecast, ci_lower, ci_upper, color='blue', alpha=0.2, label='95% Доверительный интервал')

# Названия осей и заголовок
plt.xlabel('Год')
plt.ylabel('Уровень участия в рабочей силе (%)')
plt.title('Прогноз уровня участия в рабочей силе в РФ (2024-2026) с историческими данными и доверительным интервалом')

# Отображение всех годов на оси X (с более мелким шагом)
plt.xticks(list(range(2003, 2027, 2)))  # Шаг 2 для читабельности

plt.grid(True, linestyle='--', alpha=0.6) # Сетка на графике
plt.legend() # Отображение легенды

# Показать график
plt.show()
```
